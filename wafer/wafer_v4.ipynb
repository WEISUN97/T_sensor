```python
from chip.chip_design_4.cell_T_beam_L900_500 import cell_T_beam_L900_500
from chip.chip_design_4.cell_T_beam_L700_375 import cell_T_beam_L700_375
from chip.chip_design_4.cell_T_multy_beam_L700_375 import cell_T_multy_beam_L700_375
from chip.chip_design_4.cell_T_multy_beam_L900_500 import cell_T_multy_beam_L900_500

from chip.cell_gap_test import cell_gap_test
from chip.cell_sides.cell_ultrasound import cell_ultrasound
from chip.cell_sides.cell_infrasound import cell_infrasound

wafer = gf.Component()


# layout of different cells on wafer

# cell_T_beam_L700_375
wafer.add_ref(cell_T_beam_L700_375).movey(20000)
wafer.add_ref(cell_T_beam_L700_375).movex(-20000).movey(20000)

# cell_T_beam_L900_500
wafer << cell_T_beam_L900_500
wafer.add_ref(cell_T_beam_L900_500).movex(-20000)

# cell_T_multy_beam_L700_375
wafer.add_ref(cell_T_multy_beam_L700_375).movex(-20000).movey(-20000)
# cell_T_multy_beam_L900_500
wafer.add_ref(cell_T_multy_beam_L900_500).movey(-20000)

# cell_gap_test
wafer.add_ref(cell_gap_test).movex(20000).movey(20000)
wafer.add_ref(cell_gap_test).movex(20000)
wafer.add_ref(cell_gap_test).movex(20000).movey(-20000)
wafer.add_ref(cell_gap_test).movex(-40000).movey(20000)
wafer.add_ref(cell_gap_test).movex(-40000)
wafer.add_ref(cell_gap_test).move((-40000, -20000))


# cell_infrasound
wafer.add_ref(cell_infrasound).movey(40000)
wafer.add_ref(cell_infrasound).movey(-35830)

# cell_ultrasound
wafer.add_ref(cell_ultrasound).movex(-20000).movey(40000)
wafer.add_ref(cell_ultrasound).movex(-20000).movey(-35830)

# wafer.show()
```

```python
# global chip marker
# center 12 areas
for i in range(1, 13):
    mark_text = gf.components.text(text=f"Area {i}", size=50, layer=(1, 0))
    x = -40000 + ((i-1) % 4) * 20000
    y = -30000 + (i-1) // 4 * 20000
    for j in range(4):
        row = j // 2
        col = j % 2
        (wafer << mark_text).move((x+2000+col*10000, y+2000+row*10000))
        (wafer << mark_text).move((x+2000+col*10000, y+8000+row*10000))
        (wafer << mark_text).move((x+7500+col*10000, y+8000+row*10000))
        (wafer << mark_text).move((x+7500+col*10000, y+2000+row*10000))



# edge 4 markers
for i in range(13, 17):
    k = i - 13
    mark_text = gf.components.text(text=f"Area {i}", size=50, layer=(1, 0))
    x = -20000 + (k % 2) * 20000
    y = -45830 + (k // 2) * 75830
    (wafer << mark_text).move((x+2000, y+2000))
    (wafer << mark_text).move((x+2000, y+14000))
    (wafer << mark_text).move((x+18000, y+2000))
    (wafer << mark_text).move((x+18000, y+14000))

```

```python
# grid
width = 750/2
circle = gf.components.circle(50000, layer=(10, 0))
circle_ref = wafer << circle
# horizontal lines of the grid
h_lines = [2, 8, 8, 8, 8, 8, 8, 8, 2]
x_size = [2, 1, 1, 1, 1, 1, 1, 1, 2]
y_space = [0, 15830, 10000, 10000, 10000, 10000, 10000, 10000, 15830]
x_temp = [-4, -3, -2, -1, 0, 1, 2, 3]
x_space = [[-2, 0], x_temp, x_temp, x_temp, x_temp, x_temp, x_temp, x_temp, [-2, 0]]
gap = 600 # um
y_space_temp = 0
for i in range(len(h_lines)):
    y_space_temp += y_space[i]
    for j in range(h_lines[i]):
        horizontal_line = gf.components.rectangle(size=(10000*x_size[i]-2*gap, width), layer=(3, 0))
        frame = wafer << horizontal_line
        frame.move((x_space[i][j]*10000+gap, 45830-y_space_temp-width/2))
# vertical lines of the grid
style1 = [10000] * 6
style2 = [15830] + [10000]*6 + [15830]
v_lines_len = [style1, style1, style2, style2, style2, style2, style2, style1, style1]
y_top = [20000, 20000, 30000, 30000, 30000, 30000, 30000, 20000, 20000]

for i in range(len(v_lines_len)):
    y_space_temp = y_top[i]
    for j in range(len(v_lines_len[i])):
        x_pos = -20000 + (i-2)*10000 - width/2
        vertical_line = gf.components.rectangle(size=(width, v_lines_len[i][j]-2*gap), layer=(3, 0))
        frame = wafer << vertical_line
        frame.move((x_pos,y_space_temp+gap))
        if j != len(v_lines_len[i])-1:
            y_space_temp -= v_lines_len[i][j+1]


# side grids



```

```python
mark_temp = gf.Component()
# marker
marker_1 = gf.components.rectangle(size=(1, 1000), layer=(6, 0))
marker_2 = gf.components.rectangle(size=(1000, 1), layer=(6, 0))
marker_basic = gf.Component()
marker_1_ref = marker_basic << marker_1
marker_2_ref = marker_basic << marker_2
marker_1_ref.move((-0.5, -500))
marker_2_ref.move((-500, -0.5))
marker_tri = gf.Component()
marker_tri.add_ref(marker_basic).movex(-1000)
marker_tri.add_ref(marker_basic).movey(1000)
marker_tri.add_ref(marker_basic).movey(-1000)
mark_left = mark_temp << marker_tri
mark_left.move((-40700, 0))
mark_right = mark_temp << marker_tri
mark_right.dmirror_x(0).movex(40700)
# mark position
''''
mark1: (-41700, 0), (-40700, 1000), (-40700, -1000)
mark2: (41700, 0), (40700, 1000), (40700, -1000)
'''''
# left label
label_left1 = gf.components.text("(-41700, 0)", size=50, layer=(6, 0))
label_left2 = gf.components.text("(-40700, 1000)", size=50, layer=(6, 0))
label_left3 = gf.components.text("(-40700, -1000)", size=50, layer=(6, 0))
label_left = gf.Component()
label_left.add_ref(label_left1).move((-200, 600))
label_left.add_ref(label_left2).move((700, 1600))
label_left.add_ref(label_left3).move((700, -400))
label_left_ref = mark_temp << label_left
label_left_ref.move((-41700, 0))
# right label
label_right1 = gf.components.text("(41700, 0)", size=50, layer=(6, 0))
label_right2 = gf.components.text("(40700, 1000)", size=50, layer=(6, 0))
label_right3 = gf.components.text("(40700, -1000)", size=50, layer=(6, 0))
label_right = gf.Component()
label_right.add_ref(label_right1).move((-200, 600))
label_right.add_ref(label_right2).move((-1200, 1600))
label_right.add_ref(label_right3).move((-1200, -400))
label_right_ref = mark_temp << label_right
label_right_ref.move((41700, 0))

wafer << mark_temp

# Ebeam alignment limit markers
marker_shape = gf.components.rectangle(size=(1, 1), layer=(5, 0))
(wafer << marker_shape).move((-20000, 30000-1))
(wafer << marker_shape).move((-20000, -30000))
(wafer << marker_shape).move((20000-1, 30000-1))
(wafer << marker_shape).move((20000-1, -30000))




# ATTENTION
# mirror the backside pattern for backside photolithography
wafer_backside = gf.boolean(A = wafer, B = wafer,  operation="not", layer1=(3, 0), layer2=(50, 0), layer=(4, 0))
wafer.add_ref(wafer_backside).dmirror_x(0)
wafer.show()
suffix = datetime.now().strftime("%Y%m%d_%H%M")
# wafer.write_gds(f"wafer_ID63{suffix}.gds", write_context_info=False, with_metadata=False)
```
