```python
from chip.chip_design_3.cell_web_design import cell_web_design
from chip.chip_design_3.cell_web_design_circle import cell_web_design_circle
from chip.chip_design_3.cell_mesh_design import cell_mesh_design
from chip.chip_design_3.cell_mesh_design_less import cell_mesh_design_less
from chip.chip_design_3.cell_web_design_less import cell_web_design_less
from chip.chip_design_3.cell_frame_test import cell_frame_test
from chip.chip_design_3.cell_frame_ten import cell_frame_ten

wafer = gf.Component()


# layout of different cells on wafer

# cell_mesh_design
for i in range(3):
    pos_y = (-7300+375)*i
    if i == 0:
        for j in range(6):
            pos_x_1 = j*(7300-375)
            pos_x_2 = -j*(7300-375)
            wafer.add_ref(cell_web_design_less(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
            if j !=0:
                wafer.add_ref(cell_web_design_less(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))
    else:
        for j in range(6):
            pos_x_1 = j*(7300-375)
            pos_x_2 = -j*(7300-375)
            wafer.add_ref(cell_mesh_design(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
            if j !=0:
                wafer.add_ref(cell_mesh_design(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))

# cell_mesh_design_less
for i in range(1):
    pos_y = (-7300+375)*(i+3)
    for j in range(6):
        pos_x_1 = j*(7300-375)
        pos_x_2 = -j*(7300-375)
        wafer.add_ref(cell_mesh_design_less(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_mesh_design_less(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))

# cell_mesh_design_less
for i in range(1):
    pos_y = (-7300+375)*(i+4)
    for j in range(3):
        pos_x_1 = j*(7300-375)
        pos_x_2 = -j*(7300-375)
        wafer.add_ref(cell_mesh_design_less(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_mesh_design_less(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))

# cell_frame_ten
for i in range(1):
    pos_y = (-7300+375)*(i+4)
    for j in range(3):
        j = j + 3
        if j == 3:
            beam_width=5 
            square_size=1000
        else:
            beam_width=2 
            square_size=1000
        pos_x_1 = j*(7300-375)
        pos_x_2 = -j*(7300-375)
        wafer.add_ref(cell_frame_ten(order_number=f'{i}_{j}',beam_width=beam_width, square_size=square_size)).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_frame_ten(order_number=f'{i}_-{j}',beam_width=beam_width, square_size=square_size )).move((pos_x_2, pos_y))

# cell_web_design
for i in range(2):
    pos_y = (7300-375)*(i+1)
    for j in range(6):
        pos_x_1 = j*(7300-375)
        pos_x_2 = -j*(7300-375)
        wafer.add_ref(cell_web_design(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_web_design(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))


# cell_web_design_circle
for i in range(2):
    pos_y = (7300-375)*(i+3)
    for j in range(6):
        pos_x_1 = j*(7300-375)
        pos_x_2 = -j*(7300-375)
        wafer.add_ref(cell_web_design_circle(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_web_design_circle(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))


# cell_frame_test top
for i in range(1):
    pos_y = (7300-375)*5.5
    for j in range(2):
        pos_x_1 = j*(14225-375)
        pos_x_2 = -j*(14225-375)
        wafer.add_ref(cell_frame_test(order_number=f'{i}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_frame_test(order_number=f'{i}_-{j}')).move((pos_x_2, pos_y))

# cell_frame_test bottom
for i in range(1):
    pos_y = (-7300+375)*5.5
    for j in range(2):
        pos_x_1 = j*(14225-375)
        pos_x_2 = -j*(14225-375)
        wafer.add_ref(cell_frame_test(order_number=f'{i+1}_{j}')).move((pos_x_1, pos_y))
        if j !=0:
            wafer.add_ref(cell_frame_test(order_number=f'{i+1}_-{j}')).move((pos_x_2, pos_y))
# wafer.show()
```

```python
# wafer size
circle = gf.components.circle(50000, layer=(10, 0))

circle_ref = wafer << circle
mark_temp = gf.Component()
# marker
marker_1 = gf.components.rectangle(size=(4, 1000), layer=(1, 0))
marker_2 = gf.components.rectangle(size=(1000, 4), layer=(1, 0))
marker_basic = gf.Component()
marker_1_ref = marker_basic << marker_1
marker_2_ref = marker_basic << marker_2
marker_1_ref.move((-2, -500))
marker_2_ref.move((-500, -2))
marker_tri = gf.Component()
marker_tri.add_ref(marker_basic).movex(-1000)
marker_tri.add_ref(marker_basic).movey(1000)
marker_tri.add_ref(marker_basic).movey(-1000)
mark_left = mark_temp << marker_tri
mark_left.move((-40700, 0))
mark_right = mark_temp << marker_tri
mark_right.dmirror_x(0).movex(40700)
# mark position
''''
mark1: (-41700, 0), (-40700, 1000), (-40700, -1000)
mark2: (41700, 0), (40700, 1000), (40700, -1000)
'''''
# left label
label_left1 = gf.components.text("(-41700, 0)", size=50, layer=(1, 0))
label_left2 = gf.components.text("(-40700, 1000)", size=50, layer=(1, 0))
label_left3 = gf.components.text("(-40700, -1000)", size=50, layer=(1, 0))
label_left = gf.Component()
label_left.add_ref(label_left1).move((-200, 600))
label_left.add_ref(label_left2).move((700, 1600))
label_left.add_ref(label_left3).move((700, -400))
label_left_ref = mark_temp << label_left
label_left_ref.move((-41700, 0))
# right label
label_right1 = gf.components.text("(41700, 0)", size=50, layer=(1, 0))
label_right2 = gf.components.text("(40700, 1000)", size=50, layer=(1, 0))
label_right3 = gf.components.text("(40700, -1000)", size=50, layer=(1, 0))
label_right = gf.Component()
label_right.add_ref(label_right1).move((-200, 600))
label_right.add_ref(label_right2).move((-1200, 1600))
label_right.add_ref(label_right3).move((-1200, -400))
label_right_ref = mark_temp << label_right
label_right_ref.move((41700, 0))

# mark2 = gf.boolean(A = mark_temp, B = mark_temp,  operation="not", layer1=(1, 0), layer2=(10, 0), layer=(2, 0))
# mark3 = gf.boolean(A = mark_temp, B = mark_temp,  operation="not", layer1=(1, 0), layer2=(10, 0), layer=(3, 0))
# wafer << mark3
# wafer << mark2
wafer << mark_temp


# ATTENTION
# mirror the backside pattern for backside photolithography
wafer_backside = gf.boolean(A = wafer, B = wafer,  operation="not", layer1=(3, 0), layer2=(50, 0), layer=(4, 0))
wafer.add_ref(wafer_backside).dmirror_x(0)
wafer.show()
suffix = datetime.now().strftime("%Y%m%d_%H%M")
# wafer.write_gds(f"ID74_wafer_{suffix}.gds", with_metadata=False, write_context_info=False)
wafer.write_gds(f"ID74_wafer_{suffix}.gds", with_metadata=False)
```
