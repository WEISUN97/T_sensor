```python
import sys
# sys.path.append("/Users/bubble/Desktop/Project/Infrasound Sensor/Layout/Code/acousticsensor")
sys.path.append("/Users/bubble/Desktop/Project/T_sensor/T_sensor")

import gdsfactory as gf
from Tools.geo_trans import round_corner
```

```python
# cell_T_beam_w5
L1 = [400, 900] # length of beams
w1 = 5 # width of beam
L2 = [375, 500] # length of hinges
w2 = [0.5, 0.75] # width of hinges

for i in range(len(L1)):
    for j in range(len(L2)):
        for k in range(len(w2)):
            # Ebeam lithography layer
            beam = gf.Component()
            T_hinge = gf.components.rectangle(size=(w2[k], L2[j]), layer=(5, 0))
            T_beam = gf.components.rectangle(size=(L1[i], w1), layer=(5, 0))
            T_corner = round_corner(4*w2[k], w2[k], 2*w2[k], rotation=90, layer=(5, 0))
            (beam << T_hinge).move((-w2[k]/2, 0))
            (beam << T_beam).move((-L1[i]/2, L2[j]))
            (beam << T_corner)
            (beam << T_corner).dmirror_y(L2[j]/2)
            # length mark
            file_name = f"L1_{L1[i]}_w1_{w1}_l2_{L2[j]}_w2_{w2[k]}"
            beam.write_gds(f"wafer_{file_name}.gds", write_context_info=False, with_metadata=False)
```

```python
# cell_T_Mbeam_L1_900
from itertools import count


L1 = 900 # length of beams
w1 = 5 # width of beam
gap = 50 # gap between beams
L2 = 500 # length of hinges
w2 = [0.5, 0.75, 1] # width of hinges
n_beam = [1, 3, 9]  # number of beams



for i in range(len(n_beam)):
    for j in range(len(w2)):
            block = gf.Component()
            total_height = L2*2 + n_beam[i]*w1 + (n_beam[i]-1)*gap
            # Ebeam lithography layer
            T_hinge = gf.components.rectangle(size=(w2[j], L2), layer=(5, 0))
            (block << T_hinge).move((-w2[j]/2, 0))
            (block << T_hinge).move((-w2[j]/2, total_height - L2))
            T_corner = round_corner(4*w2[j], w2[j], 2*w2[j], rotation=90, layer=(5, 0))
            (block << T_corner)
            (block << T_corner).dmirror_y(L2/2)
            (block << T_corner).dmirror_y(total_height/2)
            (block << T_corner).dmirror_y(L2/2).dmirror_y(total_height/2)



            T_beam = gf.components.rectangle(size=(L1, w1), layer=(5, 0))
            for k in range(n_beam[i]):
                (block << T_beam).move((-L1/2, L2+ k*(w1 + gap)))
            h_connector = w1*n_beam[i]+gap*(n_beam[i]-1)
            T_connector = gf.components.rectangle(size=(w1, h_connector), layer=(5, 0))
            for k in range(3):
                if k == 1:
                    (block << T_connector).move((L1/2*(k-1)-w1/2, L2))
                else:
                    (block << T_connector).move((L1/2*(k-1), L2))
            # file_name 
            file_name = f"L1_{L1}_w1_{w1}_l2_{L2}_w2_{w2[j]}_n_{n_beam[i]}"
            block.write_gds(f"wafer_{file_name}.gds", write_context_info=False, with_metadata=False)
```
